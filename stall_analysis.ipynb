# Wanjiku Gikomba Stall Analysis

```python
#python
# Provided Dataset
daily_sales = [2300, 1950, 2800, 2100, 3600, 5200, 4100,
               2500, 2200, 3100, 2400, 3900, 5600, 4700]   # 14 days, KES
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"] * 2

# qty = kg in stock, reorder = reorder level (kg), cost/price = KES per kg
stock = {
    "Tomatoes": {"qty": 18, "reorder": 25, "cost": 60, "price": 90},
    "Onions":   {"qty": 40, "reorder": 20, "cost": 70, "price": 100},
    "Sukuma":   {"qty": 9,  "reorder": 15, "cost": 30, "price": 50},
    "Potatoes": {"qty": 55, "reorder": 30, "cost": 50, "price": 75},
    "Avocado":  {"qty": 12, "reorder": 10, "cost": 80, "price": 140},
}
```

 Week-on-Week Growth

```python
#python
def week_totals(sales):
    """
    Splits 14 days of sales into two weeks using slicing and
    returns a tuple of total sales for (Week 1, Week 2).
    """
    week1_total = sum(sales[:7])
    week2_total = sum(sales[7:])
    return week1_total, week2_total

# Unpack the tuple
week1_total, week2_total = week_totals(daily_sales)

# Calculate growth percentage
wow_growth = ((week2_total - week1_total) / week1_total) * 100
```

Day Classification & Counting

```python
#python
def classify_day(amount, average):
    """
    Classifies a day's sales revenue against the period average.
    Returns 'Strong' (>20% above avg), 'Weak' (<20% below avg), or 'Normal'.
    """
    if amount > (average * 1.20):
        return "Strong"
    elif amount < (average * 0.80):
        return "Weak"
    else:
        return "Normal"

# Apply the classification and use dictionary counting pattern
day_counts = {"Strong": 0, "Normal": 0, "Weak": 0}
for sale in daily_sales:
    category = classify_day(sale, avg_sales)
    day_counts[category] = day_counts.get(category, 0) + 1
```


Weekdays vs. Weekends


```python
#python
weekday_sales = []
weekend_sales = []

for index, sale in enumerate(daily_sales):
    day_name = days[index]
    if day_name in ["Sat", "Sun"]:
        weekend_sales.append(sale)
    else:
        weekday_sales.append(sale)

avg_weekday = sum(weekday_sales) / len(weekday_sales)
avg_weekend = sum(weekend_sales) / len(weekend_sales)

# Calculate difference as a percentage of weekday average
weekend_diff_pct = ((avg_weekend - avg_weekday) / avg_weekday) * 100
```

Inventory Review and Gross Margin

```python
#python
def check_reorder_levels(stock_dict):
    """
    Evaluates stock dictionary and returns a list of product names
    whose quantity is at or below their custom reorder threshold.
    """
    items_to_reorder = []
    for item, metrics in stock_dict.items():
        if metrics["qty"] <= metrics["reorder"]:
            items_to_reorder.append(item)
    return items_to_reorder

# Fetch items to restock
restock_list = check_reorder_levels(stock)

# Calculate gross margins for all products
margins = {}
for item, metrics in stock.items():
    margin_pct = ((metrics["price"] - metrics["cost"]) / metrics["price"]) * 100
    margins[item] = margin_pct
```
